```python
from fastmcp import Client
from mcpserver import mcp
```

### Client(mcp) -> this passes the mcp server object for stdio transport

### Client("http://127.0.0.1:8000/mcp") -> this passes the server address for http transport

```python
async with Client("http://127.0.0.1:8000/mcp") as client:
    tools = await client.list_tools()


for tool in tools:
    print(f"{tool.name} - {tool.description}")
```

```python
async with Client("http://127.0.0.1:8000/mcp") as client:
    resources = await client.list_resources()


for resource in resources:
    print(f"{resource.name} - {resource.description}")

```

```python
# calling a tool

async with Client(mcp) as client:
    created_todo = await client.call_tool(
        "create_todo",
        {"title": "Buy groceries", "description": "Buy groceries from the store", "status": "pending"}
    )

    print(f"Created todo: {created_todo}")
```

```python
async with Client(mcp) as client:
    todos = await client.call_tool("list_todos")

for todo in todos.data:
    print(f"Todo: {todo}")


```
